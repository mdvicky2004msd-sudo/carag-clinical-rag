```python
import json
import sys
from IPython import get_ipython

# Path to the notebook file
notebook_path = '/content/CARAG_Clinical_Privacy_Study_Colab.ipynb'

# Load the notebook JSON structure
with open(notebook_path, 'r', encoding='utf-8') as f:
    nb_data = json.load(f)

# Filter and run all python code cells
ipython = get_ipython()
cells = nb_data.get('cells', [])
code_cells = [c for c in cells if c.get('cell_type') == 'code']

print(f"Found {len(code_cells)} code cells. Executing them in order...\n")

for idx, cell in enumerate(code_cells, 1):
    source_code = ''.join(cell.get('source', []))
    if source_code.strip():
        print(f"--- Executing Cell {idx}/{len(code_cells)} ---")
        # Execute the code block in the current global namespace
        ipython.run_cell(source_code)
```

```python
# Execute the remaining cells of the notebook (cells 8 to 11)
for idx in range(7, len(code_cells)):
    cell = code_cells[idx]
    source_code = ''.join(cell.get('source', []))
    if source_code.strip():
        print(f"--- Executing Cell {idx+1}/{len(code_cells)} ---")
        ipython.run_cell(source_code)
```
